```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
```

```python
df['IsAlone'] = df['FamilySize'].apply(lambda x: 1 if x == 1 else 0)
```

```python
df['Title'] = df['Name'].apply(lambda x: x.split(',')[1].split('.')[0].strip())
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Title', data=df, palette='Set2')
plt.title('Distribution of Titles')
plt.xticks(rotation=45)
plt.show()
```

```python
# Missing Data Heatmap
import seaborn as sns
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Data Heatmap')
plt.show()
```

```python
# Only include numeric columns for describe
df.describe(include=[np.number])  
```

```python
# Check skewness of numerical features
print(df[['Age', 'Fare']].skew())
```

```python
# Apply log transformation to Fare to reduce skewness
df['Fare'] = np.log1p(df['Fare'])
```

```python
# Convert 'Age' and 'Fare' to numeric values, coercing errors into NaN
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Fare'] = pd.to_numeric(df['Fare'], errors='coerce')

# Fill missing values (if any) with median or mean
df['Age'].fillna(df['Age'].median(), inplace=True)
df['Fare'].fillna(df['Fare'].median(), inplace=True)
```

```python
# Check data types to ensure correct conversion
print(df.dtypes)
```

```python
# Describe data with more stats, after cleaning
print(df.describe(include='all'))

# Check skewness again
print(df[['Age', 'Fare']].skew())
```

```python
# Histograms for Age and Fare to visualize their skewness
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
df['Age'].hist(bins=30, color='skyblue', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
df['Fare'].hist(bins=30, color='green', edgecolor='black')
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Apply log transformation to Fare
df['Fare'] = np.log1p(df['Fare'])

# Visualize the transformed Fare distribution
plt.figure(figsize=(6, 4))
df['Fare'].hist(bins=30, color='orange', edgecolor='black')
plt.title('Transformed Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```
